### Debt-to-Equity

$$
\text{Debt-to-Equity Ratio}
=
\frac{\text{Borrowings}}
{\text{Equity Capital} + \text{Reserves}}
$$

```python
# Sort  debt_to_equity  low to high with the comapany name and year
df_sorted = df.sort_values(by='debt_to_equity', ascending=True)
df_sorted = df_sorted[['symbol', 'year_id', 'debt_to_equity']]
# make the symbol unique and get the latest year for each symbol
df_sorted = df_sorted.drop_duplicates(subset=['symbol'], keep='last')
df_sorted.head(50)
```

```python
# from last 10 companies 
df_sorted.tail(10)
```

### Equity Ratio

$$
\text{Equity Ratio}
=
\frac{\text{Equity Capital} + \text{Reserves}}
{\text{Total Assets}}
$$


```python
# Now the same for equity_ratio
df_sorted_equity = df.sort_values(by='equity_ratio', ascending=True)
df_sorted_equity = df_sorted_equity[['symbol', 'year_id', 'equity_ratio']]
# make the symbol unique and get the latest year for each symbol
df_sorted_equity = df_sorted_equity.drop_duplicates(subset=['symbol'], keep='last')
df_sorted_equity.head(50)
```

```python
df_sorted_equity.tail(10)
```

### Book Value per Share

$$
\text{Book Value per Share}
=
\frac{\text{Equity}}
{\dfrac{\text{Equity Capital}}{\text{Face Value}}}
$$


```python
# now the same thing for Book value per share
df_sorted_book_value = df.sort_values(by='book_value_per_share', ascending=True)
df_sorted_book_value = df_sorted_book_value[['symbol', 'year_id', 'book_value_per_share']]
# make the symbol unique and get the latest year for each symbol
df_sorted_book_value = df_sorted_book_value.drop_duplicates(subset=['symbol'], keep='last')
df_sorted_book_value.head(30)
```
